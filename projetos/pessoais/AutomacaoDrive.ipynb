# Automação de Sistemas e Processos com Python

### Desafio:

Todos os dias, o nosso sistema atualiza as vendas do dia anterior.
O seu trabalho diário, como analista, é enviar um e-mail para a diretoria, assim que começar a trabalhar, com o faturamento e a quantidade de produtos vendidos no dia anterior

E-mail da diretoria: seugmail+diretoria@gmail.com<br>
Local onde o sistema disponibiliza as vendas do dia anterior: https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga

Para resolver isso, vamos usar o pyautogui, uma biblioteca de automação de comandos do mouse e do teclado


Referência do pyautogui: https://pyautogui.readthedocs.io/en/latest/quickstart.html

```python
#bibliotecas do python
#   - pyautogui -> automaçoes 
#   - pandas -> dados
#   - openpyxl -> trabalhar com bases de dados em excel 


#Comandos principais do pyautogui
#       - pyautogui.click -> clicar com o mouse 
#       - pyautogui.write -> escrever um texto 
#       - pyautogui.press -> aperta 1 tecla
#       - pyautogui.hotkey -> combinaçao de teclas 
#       - pyautogui.PAUSE -> configuraçao para definar a pause entre um comando e outro  

#time.sleep -> para dar uma pausa no codigo 
```

```python
import pyautogui
import time

# Passo a passo do desafio 
# Passo 1: Entrar no sistema da empresa (link do drive)
#    - Abrir o Chrome 
pyautogui.press('win')
time.sleep(1)
pyautogui.write('chrome')
time.sleep(1)
pyautogui.press('enter')
time.sleep(1.5)
pyautogui.click(x=413, y=540, clicks=2)
time.sleep(2)
pyautogui.hotkey("win","up")


time.sleep(2)
#    - Digitei o link do sistema (link do drive)
link = 'https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga'
pyautogui.write(link)
#    - Apertei enter e esperei 
pyautogui.press('enter')
time.sleep(7)

```

```python
# Passo 2: Navegar no sistema para encontrar a base de dados 
pyautogui.click(x=682, y=505, clicks=2)
time.sleep(3)

# Passo 3: Exportar a base de dados (baixar o arquivo)https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga
pyautogui.click(x=1261, y=509, clicks=1)
time.sleep(2)
pyautogui.click(x=1137, y=161, clicks=2)
time.sleep(6)
```

### Vamos agora ler o arquivo baixado para pegar os indicadores

- Faturamento
- Quantidade de Produtos

```python
import pandas

# Passo 4: Calcular os indicadores (faturamento e quantidade de produtos vendidos)
# abrir a base de dados 
caminho = r"C:\Users\sophi\Downloads\Vendas - Dez.xlsx"
tabela = pandas.read_excel(caminho)
# ver as informaçoes da base de dados
display(tabela)
# somar o faturamento de todos os produtos = somar a coluna de valor final 
faturamento = tabela['Valor Final'].sum()
# somar a quantidade de produtos = somar a coluna de quantidade 
qtde_produtos = tabela['Quantidade'].sum()

print(faturamento)
print(qtde_produtos)
```

### Vamos agora enviar um e-mail pelo gmail

```python
import pyperclip #copia o texto exatamento como esta com acentos e caracteres especiais

# Passo 5: Enviar as informaçoes por email 
pyautogui.click(x=290, y=21,clicks=1)
time.sleep(2)
pyautogui.click(x=1164, y=142,clicks=1)
time.sleep(6)
pyautogui.click(x=84, y=187,clicks=1)
time.sleep(2)

# para quem enviar o email 
pyautogui.click(x=855, y=309,clicks=1)
pyautogui.write('sophia.rpsouza@gmail.com')
time.sleep(2)
pyautogui.press('tab') #para selecionar o email
pyautogui.press('tab') #para passar para o campo do assunto
# qual o assunto
pyperclip.copy('Relatório de Vendas')
pyautogui.hotkey('ctrl', 'v')
pyautogui.press('tab') #para passar para o campo do corpo do email 
# qual o corpo do email
time.sleep(1)
texto =  f'''
Prezados,

Segue o relatório de vendas de hoje.
Faturamento = R$ {faturamento:,.2f}
Quantidade de produtos vendidos = {qtde_produtos:,}

Qualquer dúvida estou à disposição.
Cordialmente,
Sophia Rezende 
''' 
pyperclip.copy(texto)
pyautogui.hotkey('ctrl' , 'v')

# enviar o email 
time.sleep(1.5)
pyautogui.click(x=852, y=694, clicks=1)
```
