```python
NomeProduto = str(input('Informe o nome do produto: '))
Categoria = str(input('Informe a categoria do produto: '))
QtdeEstoque = float(input('Informe a quantidade atual do produto no estoque: '))

alimentos = 50  
bebidas = 75
limpeza = 30

if NomeProduto and Categoria and QtdeEstoque:
    if Categoria == alimentos:       
        QtdeEstoque < alimentos
        print(f'Solicitar {NomeProduto} a equipe de compras, temos apenas {QtdeEstoque} em estoque.')
    elif Categoria == bebidas:
        QtdeEstoque < bebidas
        print(f'Solicitar {NomeProduto} a equipe de compras, temos apenas {QtdeEstoque} em estoque.')
    elif Categoria == limpeza:
        QtdeEstoque < limpeza
        print(f'Solicitar {NomeProduto} a equipe de compras, temos apenas {QtdeEstoque} em estoque.')
    else:
        pass
else:
    print('Preencher as informaçoes corretamente')
```
